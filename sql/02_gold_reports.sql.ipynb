```python
%sql
CREATE OR REPLACE TABLE redcare_dev.operations.gold_daily_orders AS 
SELECT 
  order_date,
  COUNT(*) AS total_orders,
  SUM((order_amount), 2) AS total_revenue
FROM 
  redcare_dev.operations.silver_orders
GROUP BY 
  order_date
ORDER BY 
  order_date;
```

```python
%sql
CREATE OR REPLACE TABLE redcare_dev.operations.gold_status AS
SELECT
  order_date,
  order_status,
  COUNT(*) AS order_count
FROM
  redcare_dev.operations.silver_orders
GROUP BY
  order_date,
  order_status
ORDER BY
  order_date,
  order_status;
```
